在 LangChain 中，几乎所有「可运行对象」——包括模型（LLM）、链（Chain）、工具（Tool）、代理（Agent）——都遵循统一的接口规范，主要包括以下几种调用方法：

| 方法名     | 说明                             | 是否推荐 | 返回类型         |
|------------|----------------------------------|-----------|------------------|
| `invoke()` | ✅ 主力方法，一次调用一次响应      | ✅ 推荐   | 单个结果（同步） |
| `ainvoke()`| 异步调用，适合 async 场景          | ✅ 推荐   | `await` 结果     |
| `stream()` | 流式调用，适合实时输出/打印         | ✅ 推荐   | `Generator`       |
| `batch()`  | 批量调用，适合多个问题一起处理       | ✅ 推荐   | 列表结果         |
| `run()`    | 已废弃的早期简写方法               | ❌ 不推荐 | 字符串或结构体   |
| `call()`   | 已废弃的字典输入方法               | ❌ 不推荐 | 字典结构结果     |

### 2.3 LangChain 常用模块介绍

    LangChain 提供了一系列模块用于构建和优化与语言模型（如GPT系列模型）的交互过程。

        其中，
        - `PromptTemplate` 可以帮助开发者定义动态的、可重用的提示模板，通过设置模板占位符和动态插入变量来灵活构建模型输入。
        - `ResponseSchema` 模块用于清晰地定义和描述模型输出的每个数据字段，例如字段名称和含义，便于实现结构化数据的解析。
        - 而`StructuredOutputParser` 则负责组合多个 `ResponseSchema` 实例，形成完整的解析规则；它不仅能指导模型返回指定格式的数据（如JSON格式），还能自动将模型输出解析为程序可直接处理的数据结构（如Python字典）。

        通过以上三个模块的组合使用，可以显著提高与语言模型交互的规范性和便利性，帮助开发者更高效、更简洁

#### PromptTemplate.from_template 简单解释

`PromptTemplate.from_template()` 是一个快捷方法，用于**根据模板字符串自动创建提示模板（PromptTemplate）**。

它会自动识别出模板中的变量（如 `{name}`、`{text}`），不需要你手动指定变量名。

#### 📘 ChatPromptTemplate vs PromptTemplate 的区别

## ✅ 一、用途和适用场景

| 类名                   | 主要用途               | 适用模型类型           |
| -------------------- | ------------------ | ---------------- |
| `PromptTemplate`     | 构建 **单轮文本 Prompt** | 传统 LLM（如 GPT-3）  |
| `ChatPromptTemplate` | 构建 **多轮对话 Prompt** | Chat 模型（如 GPT-4） |

---

## ✅ 二、格式结构示例

### 1. `PromptTemplate`

```python
from langchain.prompts import PromptTemplate

prompt = PromptTemplate.from_template("你是谁？请用{style}风格回答")
print(prompt.format(style="幽默"))
# 输出：你是谁？请用幽默风格回答
```

* 纯字符串模板
* 不支持角色信息
* 适用于 `model.predict(prompt)` 调用

---

### 2. `ChatPromptTemplate`

```python
from langchain.prompts.chat import ChatPromptTemplate
from langchain.schema import HumanMessage, SystemMessage

chat_prompt = ChatPromptTemplate.from_messages([
    ("system", "你是一位有礼貌的AI助手"),
    ("human", "你好，请用{style}风格回答我是谁？")
])

messages = chat_prompt.format_messages(style="幽默")
print(messages)
# 输出：[SystemMessage(content=...), HumanMessage(content=...)]
```

* 基于消息结构（支持 system/human/ai 角色）
* 适用于 `chat_model.predict_messages(messages)` 调用
* 支持多轮对话、上下文信息、Memory 等功能

---

## ✅ 三、可组合性差异

| 功能           | `PromptTemplate` | `ChatPromptTemplate` |
| ------------ | ---------------- | -------------------- |
| 是否支持角色设定     | ❌                | ✅                    |
| 是否支持多轮对话     | ❌                | ✅                    |
| 是否可组合 memory | ❌                | ✅                    |
| 是否适用于 Agent  | ❌                | ✅                    |
| 使用复杂结构能力     | 弱                | 强                    |

---

## ✅ 四、总结对比表

| 比较项         | `PromptTemplate` | `ChatPromptTemplate` |
| ----------- | ---------------- | -------------------- |
| 面向模型类型      | 文本生成类（text LLM）  | Chat类模型（chat LLM）    |
| 输入结构        | 字符串模板            | 消息列表（带角色）            |
| 输出结构        | 字符串              | `Message` 对象列表       |
| 多轮对话支持      | ❌                | ✅                    |
| 是否支持 Memory | ❌                | ✅                    |
| 是否适合 Agent  | ❌                | ✅                    |

---

## ✅ 使用建议

* 如果你是在构建一个 **静态提示词（如任务指令）**，使用 `PromptTemplate` 即可。
* 如果你在开发 **多轮对话、智能体（Agent）、上下文记忆** 等功能，推荐使用 `ChatPromptTemplate`。


### prompt.format 简单解释

`prompt.format()` 是用于**给提示模板中的变量赋值**的方法。

当你使用 `PromptTemplate` 创建了一个带有占位符（如 `{name}`、`{text}`）的模板后，`format()` 可以填入这些变量，生成最终要发送给语言模型的完整提示。


#### ✅ 示例：
```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template("你好，{name}，今天是{text}")
result = prompt.format(name="小明", text="星期一")
print(result)
# 输出：你好，小明，今天是星期一

````python
from langchain_core.prompts import PromptTemplate
from langchain.output_parsers import ResponseSchema,StructuredOutputParser

schemas = [
    ResponseSchema(name="name",description="用户的姓名"),
    ResponseSchema(name="age",description="用户的年龄")
]

parser = StructuredOutputParser.from_response_schemas(schemas)

prompt = PromptTemplate.from_template(
    "请根据以下内容提取用户信息，并返回 JSON 格式：\n{input}\n\n{format_instructions}"
)

chain = (
    prompt.partial(format_instruction=parser.get_format_instructions())
    | model
    | parser

)

result = chain.invoke({"input":"用户叫李雷，今年25岁，是一名工程师。"})
print(result)
````
