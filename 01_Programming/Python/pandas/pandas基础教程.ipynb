# Pandas 基础教程

以「月度账单」为例，学习 pandas 常用操作

```python
import pandas as pd
```

## 1. 读取 Excel

`header=None` 表示没有表头行，第一行就是数据

```python
df = pd.read_excel('月度账单.xlsx', header=None)
df.columns = ['项目', '金额']  # 给列起名
df
```

## 2. 选取数据

`df['列名']` 选一列，`df.loc[行号]` 选一行

```python
# 选取金额列
amounts = df['金额']

# 总支出
total = amounts.sum()
print(f'本月总支出: {total}')

# 筛选金额大于200的
df[df['金额'] > 200]
```

## 3. 添加新数据

用 `loc` 添加一行

```python
# 添加一笔新支出
df.loc[len(df)] = ['外卖', 45]
df
```

## 4. 统计

`describe()` 快速查看统计摘要

```python
df['金额'].describe()
```

## 5. 写出 CSV

```python
df.to_csv('月度账单结果.csv', index=False, encoding='utf-8-sig')
print('✅ 已生成 月度账单结果.csv')
```
